## 4. Example Plot <a class="anchor" id="plot"></a>

### Visualization from scikit-learn

The code below uses the Iris dataset, which is available through scikit-learn. The plot shows the average sepal length for each Iris species.

```python

import matplotlib.pyplot as plt
import pandas as pd
from sklearn.datasets import load_iris
```

```python

iris = load_iris()
```

```python

iris_df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)
```

```python

iris_df["species_code"] = iris.target
```

```python

iris_df["species_name"] = iris_df["species_code"].apply(lambda x: iris.target_names[x])
```

```python
iris_df.head()
```

```python
# Calculate the average sepal length for each species
avg_sepal_length = iris_df.groupby("species_name")["sepal length (cm)"].mean()
```

```python
# Display the average sepal length values
avg_sepal_length
```

```python
# Create a bar plot of average sepal length by species
plt.figure(figsize=(8, 5))

avg_sepal_length.plot(kind="bar")

plt.title("Average Sepal Length by Iris Species")
plt.xlabel("Iris Species")
plt.ylabel("Average Sepal Length (cm)")
plt.xticks(rotation=0)

plt.show()
```

The plot shows that the average sepal length differs across the three Iris species. This is a simple example of how scikit-learn datasets can be used with Python libraries such as pandas and matplotlib to explore and visualize data.

Although this plot is simple, it demonstrates a common data analysis workflow: loading data, organizing it into a DataFrame, grouping the data, calculating a summary statistic, and visualizing the result.
